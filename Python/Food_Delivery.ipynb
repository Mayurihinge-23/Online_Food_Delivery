```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_excel('Food_Delivery.xlsx')
```

```python
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df.drop_duplicates(inplace=True)
```

```python
df.duplicated().sum()
```

```python
df.replace(['N/A','NULL',''],np.nan,inplace=True)
```

```python
df
```

```python
df=df.apply(lambda x:x.str.strip() if x.dtype=='object' else x)
```

```python
df
```

```python
df=df[df['Age']>16]
```

```python
df=df[df['Family size']>0]
```

```python
df = df.copy()
```

```python
df['Output']=df['Output'].fillna('Yes')
```

```python
df
```

```python
df['Age_Group'] = pd.cut(df['Age'],bins=[0, 25, 35, 50, 100],labels=['18-25', '26-35', '36-50', '50+'])
```

```python
df['Family_Size_Group'] = pd.cut(
    df['Family size'],
    bins=[0, 2, 4, float('inf')],
    labels=['Small', 'Medium', 'Large']
)
```

```python
df['Feedback_Flag'] = df['Feedback'].map({
    'Positive': 1,
    'Negative': 0
})
```

```python
df['Output_Flag'] = df['Output'].map({
    'Yes': 1,
    'No': 0
})
```

```python
education_map = {
    'School': 1,
    'Graduate': 2,
    'Post Graduate': 3,
    'Ph.D': 4
}

df['Education_Level'] = df['Educational Qualifications'].map(education_map)
```

```python
df
```

```python
# final data visualization
df.info()
```

```python
# export clean data
df.to_csv("Food_Clean_Data.csv",index=False)
```
